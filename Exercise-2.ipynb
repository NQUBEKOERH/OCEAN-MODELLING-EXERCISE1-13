## Exercise 2
Plotting analytical solution of the model, to show time evolution of phytoplankton and nutrient concentration
P + N =P + N
P(t) =P* P + N/P + Ne^−k′(P+N)t

```python
# importing statement 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Parameters
k = 0.1  
P0 = 0.1  
N0 = 5 

#time axes in days
t = np.linspace(0, 20, 200)

# Total concentration
total_nitrogen = P0 + N0

# Analytical solutions
P_t = P0 * (total) / (P0 + N0 * np.exp(-k * total * t))
N_t = total - P_t

```

```python
#plotting time evolution of phytoplankton and nutrients concentration
plt.plot(t, P_t, label='Phytoplankton (P)', color='green', linewidth=2)
plt.plot(t, N_t, label='Nutrient (N)', color='blue', linewidth=2)

# INSERT THE HORIZONTAL LINE HERE
plt.axhline(y=total_nitrogen, color='crimson', linestyle=':', linewidth=1.5, 
            label=f'Total Nitrogen Capacity ({total_nitrogen} mmol m$^{{-3}}$)')

plt.xlabel('Time (days)', fontsize=12)
plt.ylabel('Concentration ($\mathrm{mmol\ N\ m^{-3}}$)', fontsize=12)
plt.title('Time Evolution of Phytoplankton and Nutrient Concentrations', fontsize=14)
plt.legend(loc='best', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(0, 20)
plt.ylim(0, total + 0.5)
plt.show()
```

#### 1 Why are the curves identical in shape?
The shape are identical mirrored, because it indicates that the consumer growth is limited by the resources and the total nitrogen remain constant at 5.1mmol/m^3

#### 2. What happens when you change the value of the uptake parameter? Think about it as the percentage of food consumed per day.
When the values of the uptake parameter is changed the percentage of food consumtion is inreasing per day and the plotted graph clearly shows when the k value for both Phytoplankton and Nutrient (0.03, 0.1, 0.3)

```python
# Fixed Parameters
P0 = 0.1  
N0 = 5.0  
t = np.linspace(0, 20, 200)  #
total = P0 + N0

# Different values of the uptake parameter (k') to test
k_prime_values = [0.03, 0.1, 0.3]
colors = ["red", "green", "purple"]

# Plotting loops
for k_prime, color in zip(k_prime_values, colors):
    # Analytical solution for Phytoplankton P(t)
    numerator = total
    denominator = P0 + N0 * np.exp(-k_prime * total * t)
    P_t = P0 * (numerator / denominator)

    # Analytical solution for Nutrients N(t)
    N_t = total - P_t

    # Plot lines with specific labels for each k' value
    plt.plot(
        t,
        P_t,
        label=f"Phytoplankton (k'={k_prime})",
        color=color,
        linewidth=2,
    )
    plt.plot(
        t,
        N_t,
        label=f"Nutrient (k'={k_prime})",
        color=color,
        linestyle="--",
        linewidth=1.5,
    )

# Chart layout styling
plt.xlabel("Time (days)", fontsize=12)
plt.ylabel("Concentration ($\mathrm{mmol\ N\ m^{-3}}$)", fontsize=12)
plt.title("Effect of Varying Uptake Rate ($k'$)", fontsize=14)
plt.legend(loc="upper right", bbox_to_anchor=(1.35, 1), fontsize=9)
plt.grid(True, linestyle="--", alpha=0.5)
plt.xlim(0, 20)
plt.ylim(-0.2, total + 0.5)
plt.show()
```

#### 3 What happens if the initial biomass of phytoplankton is close to the amount of resource available? How does the shape of the curve change?
The graph below shows that if Po is 2.5 it accelerate slowly reaching an equilibrium phase.

```python

# Model parameters (P0 is chosen close to N0)
k = 0.1 
P0 = 2.5  
N0 = 2.6  
t = np.linspace(0, 20, 200)  

# Total carrying capacity (P0 + N0)
total = P0 + N0

# Analytical solution for Phytoplankton P(t)
numerator = total
denominator = P0 + N0 * np.exp(-k_prime * total * t)
P_t = P0 * (numerator / denominator)

# Analytical solution for Nutrients N(t)
N_t = total - P_t

# Plotting the results
plt.plot(t, P_t, label="Phytoplankton ($P$)", color="green", linewidth=2)
plt.plot(t, N_t, label="Nutrient ($N$)", color="blue", linewidth=2)

# Axis labels and styling
plt.xlabel("Time (days)", fontsize=12)
plt.ylabel("Concentration ($\mathrm{mmol\ N\ m^{-3}}$)", fontsize=12)
plt.title("Time Evolution when $P_0$ is Close to $N_0$", fontsize=14)
plt.legend(loc="best", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.5)
plt.xlim(0, 20)
plt.ylim(0, total + 0.5)

# Save the figure
plt.savefig("high_initial_biomass.png", bbox_inches="tight")
```
