```python
# importing statement 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Parameters
k = 0.1  
P0 = 0.1  
N0 = 5 

#time axes in days
t = np.linspace(0, 20, 200)

# Total concentration
total_nitrogen = P0 + N0

# Analytical solutions
P_t = P0 * (total) / (P0 + N0 * np.exp(-k * total * t))
N_t = total - P_t

```

```python
#plotting time evolution of phytoplankton and nutrients concentration
plt.plot(t, P_t, label='Phytoplankton (P)', color='green', linewidth=2)
plt.plot(t, N_t, label='Nutrient (N)', color='blue', linewidth=2)

# INSERT THE HORIZONTAL LINE HERE
plt.axhline(y=total_nitrogen, color='crimson', linestyle=':', linewidth=1.5, 
            label=f'Total Nitrogen Capacity ({total_nitrogen} mmol m$^{{-3}}$)')

plt.xlabel('Time (days)', fontsize=12)
plt.ylabel('Concentration ($\mathrm{mmol\ N\ m^{-3}}$)', fontsize=12)
plt.title('Time Evolution of Phytoplankton and Nutrient Concentrations', fontsize=14)
plt.legend(loc='best', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(0, 20)
plt.ylim(0, total + 0.5)
plt.show()
```

#### 1 Why are the curves identical in shape?
The shape are identical mirrored,  as phytoplankton increases, nutrients decrease by exactly the same amount, because it indicates that the consumer growth is limited by the resources and the total nitrogen remain constant at 5.1mmol/m^3

#### 2. What happens when you change the value of the uptake parameter? Think about it as the percentage of food consumed per day.
When the values of the uptake parameter is changed the percentage of food consumtion is inrceasing per day and the plotted graph clearly shows when the k value for both Phytoplankton and Nutrient (0.03, 0.1, 0.3,0.4,0.5) changes the phytoplankton increases while the nutrients decreases, which increase that the phytoplankton and consuming or uptaking the nutrients therefore there is gradual growth.

```python
# Fixed Parameters
P0 = 0.1
N0 = 5.0
t = np.linspace(0, 20, 200)
total = P0 + N0

# Different values of the uptake parameter (k') to test
k_prime_values = [0.03, 0.1, 0.3, 0.4,0.5]
colors = ["red", "green", "purple", "orange", "blue"]

# Create a figure with 1 row and 2 columns of subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plotting loops
for k_prime, color in zip(k_prime_values, colors):
    # Analytical solution for Phytoplankton P(t)
    numerator = total
    denominator = P0 + N0 * np.exp(-k_prime * total * t)
    P_t = P0 * (numerator / denominator)

    # Analytical solution for Nutrients N(t)
    N_t = total - P_t

    #Plot Phytoplankton on the 1st subplot (ax1)
    ax1.plot(
        t, P_t, label=f"Phytoplankton (k'={k_prime})", color=color, linewidth=2
    )

    # Plot Nutrients on the 2nd subplot (ax2) 
    ax2.plot(
        t,
        N_t,
        label=f"Nutrient (k'={k_prime})",
        color=color,
        linestyle="--",
        linewidth=2,
    )

# Phytoplankton Subplot
ax1.set_xlabel("Time (days)", fontsize=11)
ax1.set_ylabel("Concentration ($\mathrm{mmol\ N\ m^{-3}}$)", fontsize=11)
ax1.set_title("Phytoplankton Biomass Evolution", fontsize=12)
ax1.legend(loc="upper left", fontsize=9)
ax1.grid(True, linestyle="--", alpha=0.5)
ax1.set_xlim(0, 20)
ax1.set_ylim(-0.2, total + 0.5)

# Nutrient Subplot
ax2.set_xlabel("Time (days)", fontsize=11)
ax2.set_ylabel("Concentration ($\mathrm{mmol\ N\ m^{-3}}$)", fontsize=11)
ax2.set_title("Nutrient Drawdown Evolution", fontsize=12)
ax2.legend(loc="upper right", fontsize=9)
ax2.grid(True, linestyle="--", alpha=0.5)
ax2.set_xlim(0, 20)
ax2.set_ylim(-0.2, total + 0.5)


plt.tight_layout()
plt.show()
```
