```python
#importing libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Expansion point
x0 = 1

# Domain around x0
x = np.linspace(0.5, 1.5, 500)

# Function 1: f(x) = x/(x+1)
f1 = x/(x+1)

taylor1 = (
    x0/(x0+1)
    + (x-x0)/(x0+1)**2
    - ((x-x0)**2)/(x0+1)**3
)

print("Function 1:")
print("f(x) = x/(x+1)")
print("Taylor Series about x0 = 1:")
print("1/2 + (x-1)/4 - (x-1)^2/8 + O((x-1)^3)")
print()

# Function 2: f(x) = 2x²/x⁴ = 2/x²
f2 = 2/x**2

taylor2 = (
    2/x0**2
    - (4/x0**3)*(x-x0)
    + (6/x0**4)*(x-x0)**2
)

print("Function 2:")
print("f(x) = 2/x²")
print("Taylor Series about x0 = 1:")
print("2 - 4(x-1) + 6(x-1)^2 + O((x-1)^3)")
print()

# Function 3: f(x) = ln(x²)
f3 = np.log(x**2)

taylor3 = (
    np.log(x0**2)
    + (2/x0)*(x-x0)
    - (1/x0**2)*(x-x0)**2
)

print("Function 3:")
print("f(x) = ln(x²)")
print("Taylor Series about x0 = 1:")
print("2(x-1) - (x-1)^2 + O((x-1)^3)")
print()

# Plotting
fig, axs = plt.subplots(1, 3, figsize=(18,5))

# Function 1
axs[0].plot(x, f1, label='Original Function')
axs[0].plot(x, taylor1, '--', label='Taylor Approximation')
axs[0].scatter(x0, x0/(x0+1))
axs[0].set_title(r'$f(x)=\frac{x}{x+1}$')
axs[0].grid(True)
axs[0].legend()

# Function 2
axs[1].plot(x, f2, label='Original Function')
axs[1].plot(x, taylor2, '--', label='Taylor Approximation')
axs[1].scatter(x0, 2/x0**2)
axs[1].set_title(r'$f(x)=\frac{2}{x^2}$')
axs[1].grid(True)
axs[1].legend()

# Function 3
axs[2].plot(x, f3, label='Original Function')
axs[2].plot(x, taylor3, '--', label='Taylor Approximation')
axs[2].scatter(x0, np.log(x0**2))
axs[2].set_title(r'$f(x)=\ln(x^2)$')
axs[2].grid(True)
axs[2].legend()

plt.suptitle('Taylor Series Approximations about x₀ = 1', fontsize=14)
plt.tight_layout()
plt.show()
```
